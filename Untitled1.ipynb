```python
# Install required packages
!pip install -q timm tqdm seaborn scikit-learn

import os
import sys
import json
import time
import random
from pathlib import Path
from typing import Tuple, List, Dict

import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import GradScaler, autocast
from torch.optim.lr_scheduler import ReduceLROnPlateau
from torchvision import transforms
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix, roc_curve, auc, precision_recall_curve,
    accuracy_score, precision_score, recall_score, f1_score
)
from tqdm.notebook import tqdm
import timm

# Check GPU
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print("=" * 50)
print("BREAST CANCER DETECTION - IDC CLASSIFICATION")
print("=" * 50)
print(f"\n✓ Device: {device}")
if device == 'cuda':
    print(f"✓ GPU: {torch.cuda.get_device_name(0)}")
    print(f"✓ VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
else:
    print("⚠ WARNING: No GPU detected. Enable GPU in Runtime > Change runtime type")
```

```python
#@title ⚙️ Configuration Settings { display-mode: "form" }

#@markdown ### 📁 Paths
DATASET_PATH = "/content/breast_histopathology" #@param {type:"string"}
OUTPUT_DIR = "/content/output" #@param {type:"string"}

#@markdown ---
#@markdown ### 🔧 Training Parameters
BATCH_SIZE = 32 #@param {type:"integer"}
NUM_EPOCHS = 15 #@param {type:"integer"}
LEARNING_RATE = 0.0001 #@param {type:"number"}
IMAGE_SIZE = 224 #@param {type:"integer"}
EARLY_STOPPING_PATIENCE = 5 #@param {type:"integer"}

#@markdown ### 📊 Data Split
TRAIN_RATIO = 0.7 #@param {type:"number"}
VAL_RATIO = 0.15 #@param {type:"number"}
TEST_RATIO = 0.15 #@param {type:"number"}

# Create output directories
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(f"{OUTPUT_DIR}/evaluation", exist_ok=True)

print("\n" + "=" * 50)
print("CONFIGURATION")
print("=" * 50)
print(f"\n📁 Dataset: {DATASET_PATH}")
print(f"📁 Output:  {OUTPUT_DIR}")
print(f"\n🔧 Batch Size:   {BATCH_SIZE}")
print(f"🔧 Epochs:       {NUM_EPOCHS}")
print(f"🔧 Learning Rate: {LEARNING_RATE}")
print(f"🔧 Image Size:   {IMAGE_SIZE}x{IMAGE_SIZE}")

# Verify dataset exists
if os.path.exists(DATASET_PATH):
    folders = [f for f in os.listdir(DATASET_PATH) if os.path.isdir(os.path.join(DATASET_PATH, f))]
    print(f"\n✓ Dataset found! {len(folders)} patient folders detected.")
else:
    print(f"\n❌ Dataset not found at {DATASET_PATH}")
```

```python
class BreastHistopathologyDataset(Dataset):
    """
    Dataset for Breast Histopathology Images.

    Structure:
        /content/breast_histopathology/
        ├── 10253/
        │   ├── 0/  (Benign)
        │   └── 1/  (Malignant)
        └── ...

    Labels derived from folder names: 0 = Benign, 1 = Malignant
    """

    CLASS_NAMES = ['Benign (IDC-)', 'Malignant (IDC+)']

    def __init__(self, root_dir: str, transform=None, subset=None):
        self.root_dir = Path(root_dir)
        self.transform = transform

        if subset is not None:
            self.samples = subset
        else:
            self.samples = self._scan_dataset()

        self.class_counts = self._count_classes()

    def _scan_dataset(self):
        """Scan dataset and collect image paths with labels."""
        samples = []

        print(f"\n🔍 Scanning: {self.root_dir}")

        for patient_folder in sorted(self.root_dir.iterdir()):
            if not patient_folder.is_dir():
                continue

            # Skip IDC_regular folder if present
            if 'IDC_regular' in patient_folder.name:
                continue

            # Look for class folders (0 and 1)
            for class_folder in patient_folder.iterdir():
                if not class_folder.is_dir():
                    continue

                try:
                    label = int(class_folder.name)
                    if label not in [0, 1]:
                        continue
                except ValueError:
                    continue

                # Collect PNG images
                for img_path in class_folder.iterdir():
                    if img_path.suffix.lower() == '.png':
                        samples.append((str(img_path), label))

        print(f"✓ Found {len(samples):,} images")
        return samples

    def _count_classes(self):
        counts = {0: 0, 1: 0}
        for _, label in self.samples:
            counts[label] += 1
        return counts

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        image = Image.open(img_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        return image, label


def get_transforms(image_size=224, is_training=True):
    """Get transforms. Training includes augmentation."""
    if is_training:
        return transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.RandomHorizontalFlip(0.5),
            transforms.RandomVerticalFlip(0.5),
            transforms.RandomRotation(20),
            transforms.ColorJitter(0.2, 0.2, 0.2, 0.1),
            transforms.ToTensor(),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
        ])
    else:
        return transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.ToTensor(),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
        ])

print("✓ Dataset class defined")
```

```python
class IDCClassifier(nn.Module):
    """EfficientNetB0 with custom classifier for IDC detection."""

    def __init__(self, num_classes=2, dropout=0.3):
        super().__init__()

        # Pretrained backbone
        self.backbone = timm.create_model(
            'efficientnet_b0', pretrained=True,
            num_classes=0, global_pool='avg'
        )

        # Custom classifier
        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(self.backbone.num_features, 512),
            nn.ReLU(inplace=True),
            nn.BatchNorm1d(512),
            nn.Dropout(dropout),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        features = self.backbone(x)
        return self.classifier(features)

print("✓ Model class defined")
```

```python
print("\n" + "=" * 50)
print("LOADING DATASET")
print("=" * 50)

# Load samples
full_dataset = BreastHistopathologyDataset(DATASET_PATH)
all_samples = full_dataset.samples
labels = [l for _, l in all_samples]

print(f"\n📊 Class Distribution:")
total = len(all_samples)
for label, count in full_dataset.class_counts.items():
    print(f"   Class {label} ({full_dataset.CLASS_NAMES[label]}): {count:,} ({100*count/total:.1f}%)")

# Split data
train_samples, temp_samples, train_labels, temp_labels = train_test_split(
    all_samples, labels, test_size=(VAL_RATIO + TEST_RATIO),
    stratify=labels, random_state=42
)
val_samples, test_samples = train_test_split(
    temp_samples, test_size=TEST_RATIO/(VAL_RATIO+TEST_RATIO),
    stratify=temp_labels, random_state=42
)

print(f"\n✓ Train: {len(train_samples):,}")
print(f"✓ Val:   {len(val_samples):,}")
print(f"✓ Test:  {len(test_samples):,}")

# Create datasets
train_dataset = BreastHistopathologyDataset(DATASET_PATH, get_transforms(IMAGE_SIZE, True), train_samples)
val_dataset = BreastHistopathologyDataset(DATASET_PATH, get_transforms(IMAGE_SIZE, False), val_samples)
test_dataset = BreastHistopathologyDataset(DATASET_PATH, get_transforms(IMAGE_SIZE, False), test_samples)

# Class weights
train_counts = train_dataset.class_counts
total_train = sum(train_counts.values())
class_weights = torch.tensor([
    total_train / (2 * train_counts[0]),
    total_train / (2 * train_counts[1])
]).to(device)
print(f"\n⚖️ Class Weights: {class_weights.cpu().numpy()}")

# Data loaders
train_loader = DataLoader(train_dataset, BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True, drop_last=True)
val_loader = DataLoader(val_dataset, BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_dataset, BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

print(f"\n✓ Loaders ready: {len(train_loader)} train, {len(val_loader)} val, {len(test_loader)} test batches")
```

```python
print("\n" + "=" * 50)
print("INITIALIZING MODEL")
print("=" * 50)

model = IDCClassifier(num_classes=2, dropout=0.3).to(device)

params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\n🧠 Model: EfficientNetB0")
print(f"   Trainable params: {params:,}")

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-4)
scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=2)

print(f"\n✓ Optimizer: AdamW (lr={LEARNING_RATE})")
print(f"✓ Mixed precision: Enabled")
```

```python
def train_epoch(model, loader, criterion, optimizer, scaler):
    model.train()
    loss_sum, correct, total = 0, 0, 0

    for images, labels in tqdm(loader, desc='Training', leave=False):
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        with autocast():
            outputs = model(images)
            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        loss_sum += loss.item() * images.size(0)
        _, pred = outputs.max(1)
        total += labels.size(0)
        correct += pred.eq(labels).sum().item()

    return loss_sum / total, correct / total


@torch.no_grad()
def validate(model, loader, criterion):
    model.eval()
    loss_sum, correct, total = 0, 0, 0

    for images, labels in tqdm(loader, desc='Validating', leave=False):
        images, labels = images.to(device), labels.to(device)

        with autocast():
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss_sum += loss.item() * images.size(0)
        _, pred = outputs.max(1)
        total += labels.size(0)
        correct += pred.eq(labels).sum().item()

    return loss_sum / total, correct / total

print("✓ Training functions ready")
```

```python
from torch.cuda.amp import GradScaler
scaler = GradScaler()

print("\n" + "=" * 50)
print("🚀 TRAINING")
print("=" * 50)

history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': [], 'lr': []}
best_acc = 0
patience = 0
start = time.time()

for epoch in range(1, NUM_EPOCHS + 1):
    lr = optimizer.param_groups[0]['lr']
    print(f"\nEpoch {epoch}/{NUM_EPOCHS} | LR: {lr:.2e}")

    train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, scaler)
    val_loss, val_acc = validate(model, val_loader, criterion)
    scheduler.step(val_loss)

    history['train_loss'].append(train_loss)
    history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss)
    history['val_acc'].append(val_acc)
    history['lr'].append(lr)

    print(f"  Train: Loss={train_loss:.4f}, Acc={100*train_acc:.2f}%")
    print(f"  Val:   Loss={val_loss:.4f}, Acc={100*val_acc:.2f}%")

    if val_acc > best_acc:
        best_acc = val_acc
        patience = 0
        torch.save(model.state_dict(), f"{OUTPUT_DIR}/breast_cancer_model.pth")
        print(f"  ✓ Best model saved! ({100*best_acc:.2f}%)")
    else:
        patience += 1
        if patience >= EARLY_STOPPING_PATIENCE:
            print(f"\n⚠ Early stopping at epoch {epoch}")
            break

duration = time.time() - start
print(f"\n" + "=" * 50)
print(f"✅ TRAINING COMPLETE")
print(f"   Time: {duration/60:.1f} min")
print(f"   Best Val Acc: {100*best_acc:.2f}%")
print(f"   Model: {OUTPUT_DIR}/breast_cancer_model.pth")

# Save history
with open(f"{OUTPUT_DIR}/training_history.json", 'w') as f:
    json.dump(history, f)
```
